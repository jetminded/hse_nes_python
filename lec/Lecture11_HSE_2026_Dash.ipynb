# Фреймворк Dash

С фреймворком Dash можно создавать удобные и интерактивные веб-приложения. В отличие от библиотек `matplotlib`, `seaborn` или `plotly express`, Dash даёт возможность не только строить графики, но и объединять их в полноценные дашборды и приложения. Созданные проекты можно разместить в интернете и открыть к ним доступ пользователям.

Его особенности:

- Dash интегрируется с библиотекой Plotly, что позволяет создавать интерактивные графики.
- Компоненты Dash, написанные на Python, позволяют добавлять фильтры, кнопки, формы и прочие элементы без изучения JavaScript или HTML.
- Dash-приложения можно легко развернуть на сервере и открыть к ним доступ через интернет.

## Основные компоненты Dash


- **Серверная часть**
- **Графическая библиотека:** использует Plotly для создания интерактивных графиков.
- **Интерактивные компоненты**

## Основные пакеты Dash

1. **Dash**  
   Базовый пакет, который формирует основу приложения через объект `dash.Dash`. Это своего рода «сердце» Dash-приложения, с которого начинается его разработка.

2. **Dash Core Components**  
   Пакет включает основные интерактивные элементы, такие как выпадающие списки, ползунки и календари. Эти компоненты позволяют создать динамичный интерфейс, где пользователи могут легко фильтровать данные или управлять элементами визуализации.

3. **Dash HTML Components**  
   Пакет для работы с HTML. Упрощенно - отрисовка
   
   > **HTML (HyperText Markup Language)** — язык разметки для создания структуры веб-страниц. Каждый элемент на веб-странице: заголовок, текст, изображение или ссылка — описывается с помощью HTML-тегов. Эти теги указывают браузеру, как именно отображать контент на экране. HTML не отвечает за внешний вид элементов (стили), а только за их структуру.

4. **Dash Bootstrap Components**  
   Дополнительный пакет, который используется для оформления (шрифты, дизайн и тп)

## HTML


### Основные элементы HTML

- **Заголовки (Heading)**

  В HTML есть шесть уровней заголовков, от `<h1>` до `<h6>`. Заголовки структурируют контент и определяют иерархию информации.

```
<h1>Это главный заголовок</h1>
<h2>Это подзаголовок</h2>
<h3>Это третий уровень заголовка</h3>
```

- **Параграфы (Paragraph)**

  Элемент `<p>` используется для создания параграфов в тексте. Он автоматически добавляет отступы сверху и снизу.

```<p>Это пример параграфа текста.</p>```


- **Контейнеры (Div)**

  Элемент `<div>` — это блочный контейнер, который используется для группировки других элементов и применения стилей. `<div>` не имеет собственного стиля и служит для организации контента.
```
<div>
    <h2>Заголовок внутри div</h2>
    <p>Параграф текста внутри div.</p>
</div>
```
- **Ссылки (Links)**

  Элемент `<a>` используется для создания гиперссылок. Атрибут `href` указывает URL-адрес ссылки.

`<a href='https://www.example.com'>Перейти на Example.com</a>`

- **Изображения (Images)**

  Элемент `<img>` используется для вставки изображений. Атрибут `src` указывает путь к изображению, а атрибут `alt` предоставляет текстовое описание.

`<img src='image.jpg' alt='Описание изображения' />`


## Создание приложения


```
import dash
from dash import html

# Создание экземпляра приложения
app = dash.Dash(__name__)
```

`app = dash.Dash(__name__)` создаёт экземпляр приложения Dash. Переменная `__name__` указывает на текущий модуль.
### Макет

> **Макет** или `layout` — это структура и содержимое приложения Dash, которые определяют, какие элементы будут отображаться на странице и как они будут организованы. `Layout` задаёт основу для интерфейса, включая заголовки, текст, графики и кнопки.

В приложении Dash `layout` задаётся через атрибут `app.layout`. Это обязательный атрибут, который указывает Dash, что должно быть отображено в браузере.

```python
app.layout = html.Div([
    html.H1('Hello, World!')
])
```

**Контейнеры**

В Dash для организации содержимого часто используются контейнеры, такие как `html.Div`, который встраивает все элементы внутрь одного блока. `Div` (от division) разделяет страницу на логические части, помогает группировать элементы и управлять их стилем и расположением.

В примере `html.Div([...])` — это корневой контейнер, который оборачивает все элементы внутри приложения, позволяя легко управлять стилями и логикой интерфейса.

**Элементы внутри `layout`**

Внутри `Div` или других контейнеров можно размещать различные HTML-компоненты. В примере добавляется `html.H1('Hello, World!')`, который создаёт заголовок первого уровня с текстом `Hello, World!`.

Dash предоставляет множество HTML-компонентов, например:

- `html.H1`, `html.H2`, `html.H3` и так далее — для заголовков разного уровня.
- `html.P` — для создания параграфов текста.
- `html.Button` — для добавления кнопок.
- `html.Img` — для отображения изображений.
- `html.Div` — для создания контейнеров, где можно группировать элементы.

**Вложенные элементы**

Dash позволяет вкладывать компоненты друг в друга, чтобы создавать многоуровневую структуру. Например, можно добавить несколько заголовков, параграфов и других компонентов в `html.Div`. Каждый элемент становится дочерним для родительского контейнера, что позволяет чётко контролировать, как содержимое располагается и как компоненты влияют друг на друга.

```python
app.layout = html.Div([
    html.H1('Заголовок приложения'),
    html.Div([
        html.P('Описание приложения'),
        html.Button('Нажми меня')
    ])
])
```

## Запуск сервера
Для отображения приложения необходимо запустить сервер:

```app.run()```

Здесь `run()` запускает сервер, с помощью которого приложение будет доступно по адресу `http://127.0.0.1:8050` или `http://localhost:8050`.

Параметры `run`:

-	`host`: IP-адрес, на котором будет работать сервер. По умолчанию `127.0.0.1` (localhost). Можно установить на `'0.0.0.0'`, чтобы сделать сервер доступным из любой сети.

-	`port`: порт, на котором будет работать сервер. По умолчанию 8050. Ты можешь указать другой порт, если требуется.

-	`debug`: если установить в `True`, сервер будет работать в режиме отладки. Это позволяет автоматически его перезагружать, если нужно внести изменения в код и вывести детальные сообщения об ошибках в консоль.

-	`jupyter_mode`: если установлен в `'inline'`, приложение будет запущено в среде Jupyter и отображено непосредственно в ячейке. Другие варианты — `'external'` (для запуска в отдельной вкладке браузера) или оставить по умолчанию для стандартного запуска.


Таким образом, приведённый код создаёт простейшее веб-приложение с заголовком `'Hello, World!'`, доступное по адресу http://127.0.0.1:8050 или http://localhost:8050 в зависимости от указанных параметров.

````python
from dash import Dash, html, Input, Output, dcc
import plotly.express as px
import dash_bootstrap_components as dbc
````

````python
app = Dash(__name__) # создали объект даш
app.layout = html.Div([ # настроили внешний вид
    html.H1("Hello, World!")
])

# Запуск сервера
app.run(jupyter_mode="external") # запустили
````

## dbc.Container

**Контейнер** в контексте веб-разработки и фреймворков, таких как Dash и Bootstrap, — это элемент, который служит обёрткой для других элементов и используется для организации и структурирования контента на странице. Контейнеры помогают устанавливать отступы, выравнивание и могут управлять шириной содержимого.

Создадим контейнер с заголовком, который будет светлого цвета и расположен по центру. Для этого воспользуемся функцией `dbc.Container`.
Основные параметры `dbc.Container`:

1.	`children`
  -	**Тип:** `list` или `dash.dependency.Component`.
  - **Описание:** содержимое контейнера. Это может быть любой Dash компонент или список компонентов.
2.	`fluid`
  -	**Тип:** `bool` (по умолчанию `False`).
  -	**Описание:** если установлено в `True`, контейнер занимает всю доступную ширину экрана. Если `False`, контейнер имеет фиксированную ширину (аналог `container`).
3.	`id`
  -	**Тип:** `str`.
  -	**Описание:** уникальный идентификатор для компонента, который может использоваться для обратных вызовов (callbacks) и управления состоянием.

````python
app = Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])

app.layout = html.Div([

    # обычный Div
    html.Div([
        html.H3("Это html.Div (без отступов)")
    ], style={
        'backgroundColor': '#ffcccc',
        'border': '2px solid red',
        'padding': '15px'
    }),

    html.Br(),

    # dbc.Container
    dbc.Container([
        html.H3("Это dbc.Container (с отступами)")
    ], style={
        'backgroundColor': '#ccffcc',
        'border': '2px solid green',
        'padding': '15px'
    })
])

if __name__ == '__main__':
    app.run()
````

## `dbc.Card`

`dbc.Card` — это компонент Dash Bootstrap, который позволяет создавать карточки, чтобы отображать контент в структурированном виде

**Основные параметры `dbc.Card`**

-	`children`: содержимое карточки, которое может включать текст, изображения, кнопки и другие Dash компоненты. Например:
  - `dbc.CardHeader` — заголовок карточки, куда передаётся строка.
  - `dbc.CardBody` — тело карточки, куда передаётся текст.
  - `dbc.Button` — кнопка. Дальше мы разберём, как настроить действия после нажатия кнопки.
- `color`: указывает цвет фона карточки (например, `'primary'`, `'secondary'`, `'success'`, `'danger'`, `'warning'`, `'info'`, `'light'`, `'dark'`).
-	`outline`: если установлено в `True`, карточка будет иметь контур, а не сплошной цвет фона.
-	`className`: дополнительные классы CSS для применения кастомных стилей.

**Когда уместно пользоваться карточкой**

Если нужно:
- визуализировать какую-то статистику, например «Самый высокий рост человека — 250 см»;
- передать визуально короткую мысль с заголовком и содержанием;
- разместить текстовую информацию в прямоугольной рамке.

**Пример.** Сделаем карточку с рекламой маковых булочек. Для этого создадим карточку и передадим в неё массив элементов: заголовок и тело карточки. В тело карточки передадим содержимое и объект кнопки с текстом «Купить».

````python
app = Dash(__name__, external_stylesheets=[dbc.themes.DARKLY])

# Создание карточки
card = dbc.Card([
    dbc.CardHeader('Ищете, что съесть?'),
    dbc.CardBody(
        [
            html.P(
                'Самые лучшие на свете вкусные булочки с маком',
            ),
            dbc.Button('Купить', color='primary', href='#'),
        ]
    )
  ]
)

# Layout приложения
app.layout = dbc.Container(
    [
        html.H1('Пример использования dbc.Card'),
        card,
    ],
    fluid=True
)

if __name__ == '__main__':
    app.run(jupyter_mode='inline')
````

## Как построить адаптивную сетку с помощью `dbc.Row` и `dbc.Col`

 Они позволяют строить сетку, которая автоматически подстраивается под ширину экрана. Элементы будут по-разному комбинироваться и отображаться в зависимости от устройства — большого монитора или мобильного телефона.

### Как это работает

- `dbc.Row` создаёт строку, которая может содержать несколько колонок.
- `dbc.Col` размещается внутри `dbc.Row` и определяет, сколько места на экране будет занимать каждый элемент.

Bootstrap делит страницу на 12 равных частей. Используя этот принцип, можно задавать ширину каждой колонки с помощью параметра, например `md=4`, который указывает, что колонка занимает 4 из 12 частей экрана. Это позволяет легко выравнивать и масштабировать элементы в зависимости от доступного пространства.

### Основные параметры

**`dbc.Row`** / **`dbc.Col`**

- `children`: принимает компонент или список компонентов.
- `align`: определяет выравнивание содержимого по вертикали.

  Доступные значения:
  - `'start'`: выравнивание по верхнему краю.
  -	`'center'`: центрирование.
  -	`'end'`: выравнивание по нижнему краю.
  -	`'baseline'`: выравнивание по базовой линии текста.
  -	`'stretch'`: содержимое растягивается по высоте строки.
- `md`: задаёт ширину колонки на экранах среднего размера и выше. Например, `md=4` означает, что колонка займёт 4 из 12 доступных единиц ширины.
- `id`: уникальный идентификатор для приложения Dash.


**Пример.** Создадим простую сетку с использованием `dbc.Row` и `dbc.Col`. Для этого создадим базовую карточку:

````python
app = Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])

# Пример карточки
card = dbc.Card([
    dbc.CardHeader('Скоро дедлайн!'),
    dbc.CardBody(
        [
            html.P(
                'На самом деле не скоро',
            ),
        ]
    )
  ]
)

# Создание макета приложения с адаптивной сеткой
app.layout = dbc.Container(
    [
        dbc.Row(
            [
                dbc.Col(card, md=4),  # Каждая колонка занимает треть ширины строки
                dbc.Col(card, md=4),
                dbc.Col(card, md=4),
            ],
            align='center'  # Выравнивание колонок по вертикали
        ),
    ],
    fluid=True  # Установим fluid=True, чтобы сетка занимала всю ширину экрана
)

if __name__ == '__main__':
    app.run(debug=True)
````

### Адаптивность при изменении размеров экрана

Используя параметры, такие как `md=4`, `lg=3`, можно регулировать ширину колонок для разных размеров экрана. Например:

- `md=4` — для средних экранов и больше.
- `lg=3` — для больших экранов.

Таким образом, можно обеспечить оптимальное отображение как на мобильных устройствах, так и на десктопах.

```
dbc.Row(
    [
        dbc.Col(card, md=6, lg=4),
        dbc.Col(card, md=6, lg=4),
        dbc.Col(card, md=12, lg=4),
    ],
    align='center'
)
```

Здесь первая и вторая колонки займут по половине экрана на средних устройствах и треть — на больших. Третья колонка занимает всю ширину строки на средних устройствах и треть — на больших.

````python
app = Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])

card = dbc.Card([
    dbc.CardHeader('Скоро дедлайн!'),
    dbc.CardBody(
        [
            html.P(
                'На самом деле не скоро',
            ),
        ]
    )
  ]
)

# Создание макета приложения
app.layout = dbc.Container(
    [
        dbc.Row(
            [
                dbc.Col(card, md=4, xs=4),
                dbc.Col(card, md=4, xs=4),
            ]
        ),
         dbc.Row(
            [
                dbc.Col(card, md=4, xs=6),
                dbc.Col(card, md=4, xs=6),
            ]
        )
    ],
    fluid=True,  # Адаптивная ширина контейнера
)

# Запуск приложения
if __name__ == '__main__':
    app.run(jupyter_mode='external')
````

### `dcc.Graph`

`dcc.Graph` — служит для отображения графиков, созданных с помощью Plotly.

**Основные параметры `dcc.Graph`**

-	`id`: уникальный идентификатор графика, который позволяет взаимодействовать с ним в коллбэках.
-	`figure`: объект, который содержит данные и настройки графика, созданный с помощью Plotly.
-	`config`: словарь, который содержит дополнительные параметры конфигурации для графика, такие как возможность масштабирования, панорамирования и другие.
-	`style`: CSS-стили, которые применяются к графику. Можно использовать для задания ширины, высоты и других стилей.
-	`className`: дополнительные классы CSS для применения стилей из Bootstrap или других стилей.

`dcc.Graph` и `dbc.Card` можно разместить внутри `layout` с помощью `dbc.Container`, `dbc.Row` и `dbc.Col`.


````python
# Загрузка данных
df = px.data.iris()
fig = px.scatter(df, x='sepal_length', y='sepal_width', color='species',
                 title='Матрица рассеяния: длина и ширина чашелистков')

# Создание графика
graph = dcc.Graph(id='scatter-graph', figure=fig)

# Создание карточки для графика
graph_card = dbc.Card(
    [
        dbc.CardHeader('График зависимости чашелистиков'),
        dbc.CardBody(graph)  # Вставляем график внутрь тела карточки
    ]
)

# Создание карточки с текстом
text_card = dbc.Card(
    [
        dbc.CardHeader('Информация о данных'),
        dbc.CardBody(
            [
                html.P('Датасет содержит информацию о трёх видах ирисов: setosa, versicolor, virginica.'),
                html.P('Данные включают длину и ширину чашелистиков и лепестков.')
            ]
        )
    ]
)

# Макет приложения
app = Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])

app.layout = dbc.Container(
    [
        html.H1('Пример использования dcc.Graph и dbc.Card'),
        dbc.Row(
            [
                dbc.Col(graph_card, md=8),  # График занимает 8 из 12 частей строки
                dbc.Col(text_card, md=4)  # Карточка с текстом занимает оставшиеся 4 части
            ]
        )
    ],
    fluid=True
)

if __name__ == '__main__':
    app.run()
````

## Работа с `Input`, `Output` и `сallback`

Коллбэки отвечают за динамическое обновление элементов интерфейса *на основе действий пользователя*

### Callback в Dash

Зачем нужны коллбэки:

- **Связывают фильтр с таблицей данных.** Коллбэк может обновлять данные в таблице на основе текстового поля или выпадающего списка, показывая только записи, соответствующие выбранным фильтрам.

- **Связывают кнопки с отображением.** Можно настроить кнопку, чтобы измененить приложение (например, нажатие кнопки меняет цвет графика). Также можно настроить кнопку так, чтобы при нажатии на неё выводилось уведомление. Например, в зависимости от результата пользователь видит уведомление «Ваше письмо отправлено» или «Отправка письма не удалась».

### Основные аспекты работы с коллбэками

- **Входы (inputs) и выходы (outputs)**

  Каждый коллбэк определяет, какие компоненты будут входами и выходами.
  
  > **Входы** (англ. inputs) — это элементы, которые пользователь может изменять, например выпадающие списки, кнопки или текстовые поля. **Выходы** (англ. outputs) — это компоненты, которые необходимо обновить, например графики или текстовые метки.

- **Декоратор @app.callback**

  Коллбэки в Dash определяются с помощью декоратора `@app.callback`, который связывает компоненты через их входы и выходы. Декоратор принимает параметры, где указываются компоненты, которые будут обновляться, и компоненты, на которые нужно реагировать.

**Пример.** Создадим коллбэк, который обновляет текст в заголовке в зависимости от выбора пользователя в выпадающем списке.


````python
app = Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])

app.layout = html.Div([
    dcc.Dropdown(
        id='my-dropdown',
        options=[
            {'label': 'Роза', 'value': 'rose'},
            {'label': 'Мимоза', 'value': 'mimosa'},
        ],
        value='rose'
    ),
    html.H1(id='my-header')
])

@app.callback(
    Output('my-header', 'children'),
    Input('my-dropdown', 'value')
)
def update_header(selected_value):
    return f'Твой выбор: {selected_value}'

if __name__ == '__main__':
    app.run(jupyter_mode='inline')
````

## `dcc.Dropdown`


> `dcc.Dropdown` — это компонент библиотеки Dash Core Components, который представляет собой *выпадающий список для выбора одного или нескольких элементов.*

Этот элемент часто используется для фильтрации данных или настройки отображения на дашборде. Компонент предоставляет пользователю возможность выбора из заранее определённого списка значений.

**Основные параметры**

-	`id`: уникальный идентификатор, который связывает компонент с коллбэками и позволяет получать или изменять его состояние.
-	`options`: список опций для выбора. Каждая опция представлена в виде словаря с двумя ключами:
  - label: отображаемое название опции.
  -	value: значение, которое будет передано при выборе этой опции.
Пример:
```python
options=[
    {'label': 'Apple', 'value': 'apple'},
    {'label': 'Banana', 'value': 'banana'}
]
```
-	`value`: значение по умолчанию, которое будет выбрано при инициализации компонента. Может быть строкой (если предлагается выбор одного значения) или списком (для множественного выбора).
- `multi`: логический параметр, который определяет, можно ли выбрать несколько значений. Если `True`, то в списке можно будет выбирать несколько элементов одновременно.
- `placeholder`: текст, который отображается до того, как пользователь сделает выбор.
-	`disabled`: логический параметр, который делает выпадающий список неактивным, если установлен в `True`.
-	`clearable`: логический параметр, который включает/выключает возможность очистки выбранного значения (если включено, появляется значок очистки).

**Особенности**

-	`dcc.Dropdown` поддерживает выбор как одного, так и нескольких значений.
-	Можно динамически обновлять список опций с помощью коллбэков.

## Параметр ID

> Параметр `id` — это уникальный идентификатор компонента, который связывает его с коллбэками.

Каждый компонент Dash может иметь параметр `id`, который обеспечивает его уникальную идентификацию в приложении. Этот идентификатор необходим для интерактивных компонентов, чтобы можно было связать их с коллбэками. Если компонент не требует взаимодействия, то указывать `id` необязательно. В случае интерактивных элементов, таких как кнопки, выпадающие списки, текстовые поля, параметр `id` обязателен.

**Пример.** Для выпадающего списка задан `id` `'dropdown-example'`, а для элемента `div` — `'output-div'`. Благодаря этому в коллбэке можно указать, что элемент `'dropdown-example'` будет входом (`Input`), а элемент `'output-div'` — выходом (`Output`).

````python
app.layout = html.Div([
    dcc.Dropdown(id='dropdown-example',
                 options=[{'label': 'Option 1', 'value': '1'},
                  {'label': 'Option 2', 'value': '2'}]),
    html.Div(id='output-div')
])
````

## Сценарии использования идентификаторов

### Связь фильтра с таблицей данных

**Пример.** Пользователь вводит строку в текстовое поле, и таблица данных автоматически обновляется, показывая только записи, соответствующие введённым данным.

````python
from dash import dash_table
import pandas as pd

# Создаём пример данных
data = {
    'name': ['Иван', 'Мария', 'Петр', 'Анна', 'Сергей', 'Елена', 'Алексей', 'Татьяна'],
    'city': ['Москва', 'СПб', 'Москва', 'Казань', 'СПб', 'Москва', 'Новосибирск', 'Казань'],
    'age': [25, 30, 35, 28, 40, 32, 27, 33],
    'salary': [50000, 60000, 75000, 55000, 80000, 62000, 58000, 67000]
}
df = pd.DataFrame(data)

app = Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])


app.layout = html.Div([
    html.H1('Фильтрация данных в таблице', className='text-center mt-4 mb-4'),
    
    html.Div([
        html.Label('Поиск по городу:', className='form-label fw-bold'),
        dcc.Input(
            id='input-text',
            type='text',
            debounce=True  # Ждёт, пока пользователь закончит печатать
        ),
        ], className='mb-4'),
    
    html.Div(id='table-div'),
], className='container mt-3')


def generate_table(dataframe):
    # Если DataFrame пустой, возвращаем сообщение об отсутствии данных
    if dataframe.empty:
        return 'Нет доступных данных для отображения.'

    # Создаём таблицу Dash из DataFrame
    return dash_table.DataTable(
        data=dataframe.to_dict('records'),
        columns=[{'name': i, 'id': i} for i in dataframe.columns],
        style_table={'overflowX': 'auto'},
        style_cell={'textAlign': 'left'},
        page_size=10,
        filter_action='native',
        sort_action='native',
        editable=True,
    )

@app.callback(
    Output('table-div', 'children'),
    Input('input-text', 'value')
)
def filter_table(input_value):
    if not input_value:
        filtered_df = df
    else:
        filtered_df = df[df['city'].str.contains(input_value, case=False, na=False)]
    
    return generate_table(filtered_df)

app.run()
````

## Сборка дашборда

На основе датасета об ирисах создим дашборд, включающий фильтрацию, графики и статистику. В итоге получится полноценное приложение, которое может обрабатывать и показывать данные, которые выбрал пользователь.

````python
import dash
from dash import dcc, html
from dash.dependencies import Input, Output
import dash_bootstrap_components as dbc
import plotly.express as px

# Загрузка данных
df = px.data.iris()

# Инициализация приложения Dash
app = dash.Dash(__name__, external_stylesheets=[dbc.themes.BOOTSTRAP])

# Основные статистики
def calculate_statistics(species_filter):
    filtered_df = df[df['species'] == species_filter] if species_filter else df
    stats = {
        'Количество измерений': len(filtered_df),
        'Средняя длина лепестков': round(filtered_df['petal_length'].mean(), 2),
        'Средняя ширина лепестков': round(filtered_df['petal_width'].mean(), 2)
    }
    return stats

# Графики
def generate_graphs(species_filter):
    filtered_df = df[df['species'] == species_filter] if species_filter else df

    fig_scatter = px.scatter(
        filtered_df, x='sepal_length', y='sepal_width', color='species',
        title='Матрица рассеяния: длина и ширина чашелистков'
    )

    fig_hist_petal_length = px.histogram(
        filtered_df, x='petal_length', color='species',
        title='Распределение длины лепестков'
    )

    fig_hist_petal_width = px.histogram(
        filtered_df, x='petal_width', color='species',
        title='Распределение ширины лепестков'
    )

    return fig_scatter, fig_hist_petal_length, fig_hist_petal_width

# Макет приложения
app.layout = dbc.Container([
    html.H1('Дашборд данных ирисов', className='text-center text-primary mb-4'),

    dbc.Row([
        dbc.Col([
            dbc.Card([
                dbc.CardHeader('Фильтр по видам'),
                dbc.CardBody([
                    dcc.Dropdown(
                        id='species-dropdown',
                        options=[{'label': species, 'value': species} for species in df['species'].unique()] + [{'label': 'Все виды', 'value': ''}],
                        value='',
                        placeholder='Выберите вид ирисов'
                    )
                ])
            ])
        ], md=4),

        dbc.Col([
            dbc.Card([
                dbc.CardHeader('Основные статистики'),
                dbc.CardBody([
                    html.P(id='measurement-count', className='card-text'),
                    html.P(id='avg-petal-length', className='card-text'),
                    html.P(id='avg-petal-width', className='card-text'),
                ])
            ])
        ], md=8)
    ], className='mb-4'),

    dbc.Row([
        dbc.Col(dcc.Graph(id='scatter-graph'), md=12)
    ]),

    dbc.Row([
        dbc.Col(dcc.Graph(id='hist-petal-length'), md=6),
        dbc.Col(dcc.Graph(id='hist-petal-width'), md=6),
    ]),

], fluid=True)

# Коллбэки для обновления статистики и графиков
@app.callback(
    [Output('measurement-count', 'children'),
     Output('avg-petal-length', 'children'),
     Output('avg-petal-width', 'children'),
     Output('scatter-graph', 'figure'),
     Output('hist-petal-length', 'figure'),
     Output('hist-petal-width', 'figure')],
    [Input('species-dropdown', 'value')]
)
def update_dashboard(species_filter):
    # Обновление статистик
    stats = calculate_statistics(species_filter)
    measurement_count = f"Количество измерений: {stats['Количество измерений']}"
    avg_petal_length = f"Средняя длина лепестков: {stats['Средняя длина лепестков']} см"
    avg_petal_width = f"Средняя ширина лепестков: {stats['Средняя ширина лепестков']} см"

    # Обновление графиков
    fig_scatter, fig_hist_petal_length, fig_hist_petal_width = generate_graphs(species_filter)

    return measurement_count, avg_petal_length, avg_petal_width, fig_scatter, fig_hist_petal_length, fig_hist_petal_width

# Запуск приложения
if __name__ == '__main__':
    app.run()
````
